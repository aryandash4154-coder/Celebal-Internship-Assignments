## Q1. Explain the roles of the Driver, Cluster Manager, and Executor in a Spark application.

### Driver
- The Driver is the main program that starts the Spark application.
- It creates the SparkSession and converts the user code into jobs and tasks.
- It builds the Directed Acyclic Graph (DAG) for execution.
- It communicates with the Cluster Manager and Executors.

### Cluster Manager
- The Cluster Manager allocates resources (CPU and memory) to the Spark application.
- It starts and manages Executor processes on worker nodes.
- Examples include Standalone, YARN, Mesos, and Kubernetes.

### Executor
- Executors are worker processes that execute the tasks assigned by the Driver.
- They store data in memory or disk during execution.
- They return the results of completed tasks back to the Driver.

## Q2. How does Spark's Lazy Evaluation strategy improve performance?

Spark uses Lazy Evaluation, which means transformations are not executed immediately.

Instead, Spark records all transformations and creates a Directed Acyclic Graph (DAG). The execution starts only when an Action such as `show()`, `count()`, or `collect()` is called.

### Advantages
- Reduces unnecessary computations.
- Optimizes the execution plan.
- Minimizes disk I/O.
- Improves execution speed for large datasets.

## Q4. Difference between CSV and Parquet

| CSV | Parquet |
|-----|----------|
| Row-based storage | Columnar storage |
| Larger file size | Compressed and smaller |
| Slower reading | Faster reading |
| Schema is not stored | Schema is stored |
| No Predicate Pushdown | Supports Predicate Pushdown |

### Why Parquet is Faster?

Since Parquet stores data column-wise, Spark reads only the required columns instead of the entire file. This reduces disk I/O, memory usage, and improves query performance.

## Q7. How does Spark use the Lineage Graph (DAG) to provide fault tolerance?

Spark maintains a Lineage Graph (Directed Acyclic Graph) that records all transformations applied to the data.

If an Executor or worker node fails, Spark does not reload the entire dataset. Instead, it uses the DAG to recompute only the lost partitions from the original data.

This makes Spark fault tolerant without requiring every intermediate result to be stored on disk.

## Q8. Filter rows where the status is 'Completed' and the amount is greater than 1000.

```python
completed_orders = df.filter(
    (col("status") == "Completed") &
    (col("amount") > 1000)
)

completed_orders.show()
```

## Q9. Explain the concept of Predicate Pushdown in Parquet and how it affects the amount of data loaded into memory.

Predicate Pushdown is an optimization technique used by Spark with Parquet files. When a filter condition is applied, Spark pushes the condition down to the Parquet file instead of reading the entire dataset.

As a result, only the required rows and columns are loaded into memory.

### Advantages
- Reduces disk I/O.
- Improves query performance.
- Reduces memory usage.
- Speeds up data processing.

## Q11. Difference between Transformations and Actions

### Transformations
Transformations create a new DataFrame from an existing one but do not execute immediately because Spark uses Lazy Evaluation.

**Examples:**
- `select()`
- `filter()`

### Actions
Actions trigger the execution of all pending transformations and produce a result.

**Examples:**
- `show()`
- `count()`

## Q15. Why is it safer to use `.show(5)` instead of `.collect()` on a multi-terabyte dataset?

- `.show(5)` displays only the first five rows and keeps the remaining data distributed across the cluster.
- `.collect()` retrieves the entire dataset to the Driver's memory.
- On very large datasets, `.collect()` can consume excessive memory and may cause an OutOfMemoryError.
- Therefore, `.show(5)` is recommended for exploring large datasets.
